```python
## 1.  list creation(5 integer values and 5 string values)

age_list = [24, 25, 26, 27, 28]

name_list = ["Arun", "Bala", "Chitra", "Divya", "Kumar"]

print("Age List:", age_list)
print("Name List:", name_list)
```

```python
## a. Append the string "Yazhini" to name_list.

name_list = ["Arun", "Bala", "Chitra", "Divya", "Kumar"]

name_list.append("Yazhini")

print(name_list)
```

```python
## 2.   list operations / Modifications

## c. Remove the string "Yazhini" from name_list.

name_list.remove("Yazhini")

print("After Remove:", name_list)
```

```python
# b. Insert the element 30 at index 2 in age_list.  

age_list = [24, 25, 26, 27, 28]

age_list.insert(2, 30)

print("After Insert:", age_list)
```

```python
# d. Pop the last element from age_list.  

age_list.pop()

print("After Pop:", age_list)
```

```python
# e. Extend the age_list with additional ages [29, 30, 26].

age_list = [24,25,26,27,28]

age_list.extend([29,30,26])

print(age_list)
```

```python
# f. Sort age_list in descending order.

age_list.sort(reverse=True)

print(age_list)
    
```

```python
#g. Find Max age, Min age and sum of all ages from age_list.

print("Maximum age:", max(age_list))
```

```python
print("Minimum age:", min(age_list))
```

```python
print("Sum of all ages:", sum(age_list))
```

```python
# Dictionary (Creation, Modification and Access):  
# a. Create a dictionary named student_marks  five students to their marks.

students_marks = {
    "Arun": 85,
    "Bala": 90,
    "Chitra": 72,
    "Divya": 88,
    "Kumar": 92
}

print(students_marks)

    
```

```python
# b. Access and print the mark of a specific student

print(students_marks["Divya"])
```

```python
# c. Add a new student "Janani" with a mark of 80 to the student_marks dictionary.

students_marks["Janani"] = 80

print(students_marks)
```

```python
# d. Update the mark of any one older student to 82.

students_marks["Divya"] = 82

print(students_marks)
```

```python
# e. Use keys(), values(), and items() methods.
print ("keys:", students_marks.keys())
```

```python
print("Items:", students_marks.items())
```

```python
# c. Create two sets:  

set1 = {1, 3, 5, 7, 9}  
set2 = {2, 3, 5, 8, 10} 

print(set1)
print(set2)
```

```python
# Operators & Conditional Statements : 
#(IF, ELIF, ELSE) 
#Performance Category Program

score = int(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score. Please enter between 0 and 10.")

elif score > 7:
    print("Performance Category: Above Average")
    print("Excellent! Keep up the good work.")

elif score >= 4:
    print("Performance Category: Average")
    print("Good effort! Keep practicing to improve.")

else:
    print("Performance Category: Below Average")
    print("Need to improve your performance. Consistent practice will lead to better results.")
```
